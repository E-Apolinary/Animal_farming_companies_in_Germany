```python
#inspecting the data in the discovery_candidates table
import pandas as pd

df = pd.read_csv(
    "D:\\ANIMAL_FARMING_GERMANY\\data\\raw\\dvt_companies.csv" #adding \\ to escape the backslash in the file path
)

print("Companies:", len(df))

display(df.head(10))

print("\nMissing values:")
print(df.isna().sum())

print("\nDuplicate companies:")
print(df.duplicated(
    subset=["company", "address"]
).sum())

print("\nCities:")
print(df["city"].value_counts().head(20))
```

```python
#checking  city quality 
# Missing cities
missing_cities = df[
    df["city"].isna() |
    (df["city"].astype(str).str.strip() == "")
]

print("Missing cities:", len(missing_cities))

# Inspect city names
print("\nCity values:")
print(df["city"].value_counts().head(30))

# Detect suspicious city values
suspicious_cities = df[
    df["city"].astype(str).str.contains(
        r"\d|Tel:|Telefon|www\.|@",
        case=False,
        regex=True,
        na=False
    )
]

print("\nSuspicious cities:", len(suspicious_cities))

display(suspicious_cities[
    ["company", "city", "address"]
].head(20))
```

```python
#investigate missing website 
missing_websites = df[
    df["website"].isna() |
    (df["website"].astype(str).str.strip() == "") # Filter out empty strings
]

print("Missing websites:", len(missing_websites))

display(missing_websites[
    ["company", "city", "website"]
].head(20))
```

```python
#checking for duplicates  
duplicates = df[
    df.duplicated(
        subset=["company", "address"],
        keep=False #marks all duplicates as True, not just the first or last occurrence
    )
]

print("Duplicate records:", len(duplicates))

display(duplicates[
    ["company", "city", "address"]
].head(20))
```

```python
#Clean the city column 
import re #useful for removing unwanted text patterns from city names, such as phone numbers or website labels

df = pd.read_csv("D:\\ANIMAL_FARMING_GERMANY\\data\\raw\\dvt_companies.csv")

def clean_city(city): #function to clean city names by removing unwanted text patterns
    if pd.isna(city): #check if the city value is NaN (missing)
        return city #if the city value is NaN (missing), return it as is 

    city = str(city).strip() #convert the city value to a string and remove leading/trailing whitespace

    # Remove telephone numbers and website labels
    city = re.split(
        r"\s+(?:Tel\.?|Telefon|zur Website|Website)\b"
        r"|\s+\+49\b", #this regex pattern matches common telephone number prefixes and website labels in German, such as "Tel.", "Telefon", "zur Website", "Website", and the country code "+49"
        city,
        maxsplit=1, #only split at the first occurrence of the pattern
        flags=re.IGNORECASE
    )[0]

    return city.strip(" ,;") #remove any leading/trailing spaces, commas, or semicolons from the cleaned city name

df["city"] = df["city"].apply(clean_city) #apply the cleaning function to the city column

#verify the cleaning results
suspicious = df[
    df["city"].str.contains(
        r"Tel:|Telefon|\+49|zur Website",
        case=False,
        regex=True,
        na=False
    )
]

print("Suspicious cities remaining:", len(suspicious))
```

```python
#updating the database
import sqlite3

db_path = "data/database/animal_farming_germany_v2.db"

with sqlite3.connect(db_path) as conn:

    updates = [
        (row.city, row.company, row.address)
        for row in df.itertuples(index=False)
    ]

    conn.executemany("""
        UPDATE discovery_candidates
        SET city = ?
        WHERE company = ?
        AND address = ?
        AND source_name = 'DVT Member Directory'
    """, updates)

    print("Rows updated:", conn.total_changes)
```
